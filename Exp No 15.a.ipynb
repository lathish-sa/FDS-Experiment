```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'Germany', 'France', 'Spain', 'France', np.nan, 'France'],
    'Age': [44.0, 27.0, 30.0, 38.0, 40.0, 35.0, np.nan, 48.0, 50.0, 37.0],
    'Salary': [72000.0, 48000.0, 54000.0, 61000.0, np.nan, 58000.0, 52000.0, 79000.0, 83000.0, 67000.0],
    'Purchased': ['No', 'Yes', 'No', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
print(df)
```

```python
df['Country'].fillna(df['Country'].mode()[0], inplace=True)
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Salary'].fillna(round(df['Salary'].mean()), inplace=True)
```

```python
country_dummies = pd.get_dummies(df['Country'], dtype=int) # Added dtype=int to get 1 and 0 instead of True/False
updated_dataset = pd.concat([country_dummies, df.iloc[:, [1, 2, 3]]], axis=1)

print("\n--- Updated Info ---")
df.info()
```

```python
updated_dataset['Purchased'] = updated_dataset['Purchased'].replace({'No': 0, 'Yes': 1})

print("\n--- Final Cleaned & Encoded Dataset ---")
print(updated_dataset)
```
