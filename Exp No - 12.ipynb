```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
array = np.array([27, 50, 44, 6, 58, 61, 23, 86, 67, 20, 75, 7, 79, 61, 90, 54])

print("Mean:", array.mean())
print("25th Percentile (Q1):", np.percentile(array, 25))
print("50th Percentile (Median):", np.percentile(array, 50))
print("75th Percentile (Q3):", np.percentile(array, 75))
print("100th Percentile (Max):", np.percentile(array, 100))
```

```python
def outDetection(arr):
    Q1, Q3 = np.percentile(arr, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)
    return lr, ur
```

```python
lr, ur = outDetection(array)
print(f"\nLower Range (lr): {lr}")
print(f"Upper Range (ur): {ur}")
```

```python
sns.displot(array, kde=True)
plt.title("Distribution of Array Data")
plt.show()
```
