```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = {
    'total_bill': [16.99, 10.34, 21.01, 23.68, 24.59],
    'tip': [1.01, 1.66, 3.50, 3.31, 3.61],
    'sex': ['Female', 'Male', 'Male', 'Male', 'Female'],
    'smoker': ['No', 'No', 'No', 'No', 'No'],
    'day': ['Sun', 'Sun', 'Sun', 'Sun', 'Sun'],
    'time': ['Dinner', 'Dinner', 'Dinner', 'Dinner', 'Dinner'],
    'size': [2, 3, 3, 2, 4]
}

tips = pd.DataFrame(data)
print(tips.head())
```

```python
sns.displot(tips.total_bill, kde=True)
plt.show()
```

```python
sns.jointplot(x=tips.tip, y=tips.total_bill)
plt.show()
```

```python
sns.jointplot(x=tips.tip, y=tips.total_bill, kind="reg")
plt.show()
```

```python
sns.jointplot(x=tips.tip, y=tips.total_bill, kind="hex")
plt.show()
```

```python
sns.pairplot(tips)
plt.show()
```

```python
print("\n--- Time Value Counts ---")
print(tips.time.value_counts())
```

```python
sns.pairplot(tips, hue='time')
plt.show()
```

```python
sns.pairplot(tips, hue='day')
plt.show()
```

```python
sns.heatmap(tips.corr(numeric_only=True), annot=True)
plt.show()
```

```python
sns.boxplot(x=tips.total_bill)  # Fixed: Explicitly added x=
plt.show()
```

```python
sns.boxplot(x=tips.tip)         # Fixed: Explicitly added x=
plt.show()
```

```python
sns.countplot(x=tips.day)       # Fixed: Explicitly added x=
plt.show()
```

```python
sns.countplot(x=tips.sex)       # Fixed: Explicitly added x=
plt.show()
```

```python
print("\n--- Plotting Pie and Bar Charts ---")
tips.sex.value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.show()
```

```python
tips.sex.value_counts().plot(kind='bar')
plt.show()
```

```python
dinner_tips = tips[tips.time == 'Dinner']
sns.countplot(data=dinner_tips, x='day')
plt.title("Dinner Counts by Day")
plt.show()
```
