```python
import numpy as np
```

```python
print("--- Array Generation ---")
array = np.random.randint(1, 100, 9)
print("Original 1D Array:\n", array)
```

```python
print("\n--- Math & Properties ---")
print("Square Root of Array:\n", np.sqrt(array))
print("\nDimensions of original array:", array.ndim)
```

```python
print("\n--- Reshaping ---")
new_array = array.reshape(3, 3)
print("\nReshaped 3x3 Matrix (new_array):\n", new_array)
print("\nDimensions of new matrix:", new_array.ndim)

print("Flattened Matrix (ravel):\n", new_array.ravel())

newm = new_array.reshape(3, 3)
print("\nMatrix (newm):\n", newm)
```

```python
# 4. Advanced Slicing Operations
print("\n--- Matrix Slicing ---")

print("newm[2, 1:3]:\n", newm[2, 1:3])

print("\nnewm[1:2, 1:3]:\n", newm[1:2, 1:3])

print("\nnew_array[0:3, 0:0]:\n", new_array[0:3, 0:0])

print("\nnew_array[0:2, 0:1]:\n", new_array[0:2, 0:1])

print("\nnew_array[0:3, 0:1]:\n", new_array[0:3, 0:1])

print("\nnew_array[1:3]:\n", new_array[1:3])
```
