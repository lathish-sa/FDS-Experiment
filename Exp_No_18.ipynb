```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

```

```python
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=["sepal.length", "sepal.width", "petal.length", "petal.width"])

target_mapping = {0: "Setosa", 1: "Versicolor", 2: "Virginica"}
df["variety"] = [target_mapping[t] for t in iris.target]

features = df.iloc[:, :-1].values
label = df.iloc[:, 4].values

xtrain, xtest, ytrain, ytest = train_test_split(
    features, label, test_size=0.2, random_state=42)

model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(xtrain, ytrain)

print("--- Model Accuracy ---")
print(f"Train Score: {model_KNN.score(xtrain, ytrain):.4f}")
print(f"Test Score: {model_KNN.score(xtest, ytest):.4f}\n")

print("--- Confusion Matrix ---")
print(confusion_matrix(label, model_KNN.predict(features)))
print("\n")

print("--- Classification Report ---")
print(classification_report(label, model_KNN.predict(features)))
```
