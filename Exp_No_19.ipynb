```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split

```

```python
np.random.seed(42)
mock_age = np.random.randint(18, 60, size=100)
mock_salary = np.random.randint(15000, 150000, size=100)

probabilities = 1 / (1 + np.exp(-(0.1 * mock_age + 0.00002 * mock_salary - 5)))
mock_purchased = np.where(probabilities > np.random.rand(100), 1, 0)

df = pd.DataFrame(
    {
        "User ID": np.random.randint(15500000, 15900000, size=100),
        "Gender": np.random.choice(["Male", "Female"], size=100),
        "Age": mock_age,
        "EstimatedSalary": mock_salary,
        "Purchased": mock_purchased,
    }
)

features = df.iloc[:, [2, 3]].values
label = df.iloc[:, 4].values

```

```python
print("--- Finding splits where Test Score > Train Score ---")
for i in range(1, 101):
    # FIXED: Added missing test_size and dynamically bound random_state to i
    x_train, x_test, y_train, y_test = train_test_split(
        features, label, test_size=0.2, random_state=i
    )

    model = LogisticRegression()
    model.fit(x_train, y_train)

    train_score = model.score(x_train, y_train)
    test_score = model.score(x_test, y_test)

    if test_score > train_score:
        print(f"Test: {test_score:.2f} | Train: {train_score:.2f} | Random State: {i}")

print("\n--- Training Final Model ---")
x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=42
)

```

```python
finalModel = LogisticRegression()
finalModel.fit(x_train, y_train)

print(f"Final Train Score: {finalModel.score(x_train, y_train):.4f}")
print(f"Final Test Score: {finalModel.score(x_test, y_test):.4f}\n")

print("--- Classification Report ---")
print(classification_report(label, finalModel.predict(features)))
```
