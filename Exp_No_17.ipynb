```python
import pickle
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
np.random.seed(42)
mock_experience = np.random.uniform(1.1, 10.5, 30)
mock_salary = (9000 * mock_experience + 28000 + np.random.normal(0, 5000, 30)).astype(int)
df = pd.DataFrame({"YearsExperience": mock_experience, "Salary": mock_salary})


df.dropna(inplace=True)

features = df.iloc[:, [0]].values
label = df.iloc[:, [1]].values

x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=42
)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)

print(f"Train Score (R2): {model.score(x_train, y_train):.4f}")
print(f"Test Score (R2): {model.score(x_test, y_test):.4f}")
print(f"Model Coefficient (Slope): {model.coef_[0][0]:.2f}")
print(f"Model Intercept: {model.intercept_[0]:.2f}\n")
```

```python
pickle.dump(model, open("SalaryPred.model", "wb"))

loaded_model = pickle.load(open("SalaryPred.model", "rb"))

try:
    yr_of_exp_input = input("Enter Years of Experience: ")
    yr_of_exp = float(yr_of_exp_input) if yr_of_exp_input else 5.0
except (ValueError, EOFError):
    yr_of_exp = 5.0
    print(f"Using default experience: {yr_of_exp} years")

yr_of_exp_NP = np.array([[yr_of_exp]])
predicted_salary = loaded_model.predict(yr_of_exp_NP)

print(f"Estimated Salary for {yr_of_exp} years of experience is: ${predicted_salary[0][0]:,.2f}")
```
